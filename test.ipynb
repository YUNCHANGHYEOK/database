```python
# OpenCV Bar Chart - Recent 30 Days Stock Data
import cv2
import numpy as np
import pandas as pd

# Read Samsung Electronics stock data
df = pd.read_csv("data/삼성전자.csv")

# Convert Close column to numeric (fix string issue)
df['Close'] = pd.to_numeric(df['Close'], errors='coerce')

# Remove missing values
df = df.dropna(subset=['Close'])

# Get recent 30 days closing price data (explicitly convert to float)
recent_data = df['Close'].tail(30).values.astype(float)
dates = df.index[-30:]

print(f"Data type: {recent_data.dtype}")
print(f"Recent 30 days data shape: {recent_data.shape}")

# Image size settings
img_width = 1200  # Increased width for 30 bars
img_height = 600
img = np.zeros((img_height, img_width, 3), dtype=np.uint8)
img[:] = (255, 255, 255)  # White background

# Bar chart settings
bar_width = 30  # Reduced bar width for more bars
margin = 60
graph_width = img_width - 2 * margin
graph_height = img_height - 2 * margin

# Data normalization (fit to graph height) - now numeric operations possible
max_value = float(np.max(recent_data))
min_value = float(np.min(recent_data))
print(f"Max value: {max_value} (type: {type(max_value)})")
print(f"Min value: {min_value} (type: {type(min_value)})")

# Prevent division by zero
if max_value == min_value:
    normalized_data = np.full(len(recent_data), graph_height / 2)
else:
    normalized_data = (recent_data - min_value) / (max_value - min_value) * graph_height

# Draw bar chart
for i, (value, norm_value) in enumerate(zip(recent_data, normalized_data)):
    x = margin + i * (graph_width // 30)  # Changed to 30 bars
    y_bottom = img_height - margin
    y_top = int(y_bottom - norm_value)
    
    # Draw bar (orange color)
    cv2.rectangle(img, (x, y_top), (x + bar_width, y_bottom), (0, 165, 255), -1)
    
    # Draw bar border
    cv2.rectangle(img, (x, y_top), (x + bar_width, y_bottom), (0, 0, 0), 1)
    
    # Show value (only every 5th bar to avoid crowding)
    if i % 5 == 0:
        text = f"{value:.0f}"
        text_size = cv2.getTextSize(text, cv2.FONT_HERSHEY_SIMPLEX, 0.4, 1)[0]
        text_x = x + (bar_width - text_size[0]) // 2
        text_y = y_top - 5
        cv2.putText(img, text, (text_x, text_y), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0, 0, 0), 1)
    
    # Show date (only every 5th day to avoid crowding)
    if i % 5 == 0:
        date_text = f"D{i+1}"
        date_text_size = cv2.getTextSize(date_text, cv2.FONT_HERSHEY_SIMPLEX, 0.3, 1)[0]
        date_x = x + (bar_width - date_text_size[0]) // 2
        date_y = y_bottom + 15
        cv2.putText(img, date_text, (date_x, date_y), cv2.FONT_HERSHEY_SIMPLEX, 0.3, (0, 0, 0), 1)

# Draw axes
# Y-axis
cv2.line(img, (margin, margin), (margin, img_height - margin), (0, 0, 0), 2)
# X-axis
cv2.line(img, (margin, img_height - margin), (img_width - margin, img_height - margin), (0, 0, 0), 2)

# Add title
title = "Samsung Electronics - Recent 30 Days Stock Price"
title_size = cv2.getTextSize(title, cv2.FONT_HERSHEY_SIMPLEX, 0.7, 2)[0]
title_x = (img_width - title_size[0]) // 2
cv2.putText(img, title, (title_x, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 0), 2)

# Add Y-axis labels
y_labels = np.linspace(min_value, max_value, 6)
for i, label in enumerate(y_labels):
    y_pos = img_height - margin - int(i * graph_height / 5)
    label_text = f"{label:.0f}"
    cv2.putText(img, label_text, (10, y_pos + 5), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0, 0, 0), 1)
    # Draw grid lines
    cv2.line(img, (margin, y_pos), (img_width - margin, y_pos), (200, 200, 200), 1)

# Save image
cv2.imwrite('data/stock_bar_chart_30days.png', img)

# Display image (in Jupyter)
from PIL import Image
import matplotlib.pyplot as plt

# Convert OpenCV image to RGB for matplotlib display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(15, 8))
plt.imshow(img_rgb)
plt.axis('off')
plt.title('OpenCV Generated Bar Chart - 30 Days')
plt.tight_layout()
plt.show()

print("Bar chart saved as 'data/stock_bar_chart_30days.png'")
print(f"Recent 30 days stock data: {len(recent_data)} data points")
print(f"Price range: {min_value:.2f} - {max_value:.2f}")
```
